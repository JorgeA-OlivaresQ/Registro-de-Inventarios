# 1. Módulo de autenticación de usuarios

## Este módulo permite registrar usuarios, iniciar sesión y validar el tipo de usuario.

```python
class Usuario:
    """
    Clase que representa a un usuario del sistema.
    """

    def __init__(self, username, password, rol):
        self.username = username
        self.password = password
        self.rol = rol


class AutenticacionUsuarios:
    """
    Clase que administra el registro, inicio de sesión
    y control de acceso según el rol del usuario.
    """

    def __init__(self):
        self.usuarios = {}
        self.usuario_actual = None

    def registrar_usuario(self, username, password, rol):
        try:
            if username in self.usuarios:
                raise ValueError("El nombre de usuario ya existe.")

            if rol not in ["administrador", "almacenero", "consulta"]:
                raise ValueError("Rol no válido.")

            self.usuarios[username] = Usuario(username, password, rol)
            print("Usuario registrado correctamente.\n")

        except ValueError as error:
            print(f"Error al registrar usuario: {error}\n")

    def iniciar_sesion(self, username, password):
        try:
            if username not in self.usuarios:
                raise ValueError("El usuario no existe.")

            usuario = self.usuarios[username]

            if usuario.password != password:
                raise ValueError("Contraseña incorrecta.")

            self.usuario_actual = usuario
            print(f"Bienvenido, {usuario.username}. Rol: {usuario.rol}\n")

        except ValueError as error:
            print(f"Error al iniciar sesión: {error}\n")

    def cerrar_sesion(self):
        if self.usuario_actual:
            print(f"Sesión cerrada para {self.usuario_actual.username}.\n")
            self.usuario_actual = None
        else:
            print("No hay ninguna sesión activa.\n")

    def verificar_acceso(self, roles_permitidos):
        if self.usuario_actual is None:
            print("Acceso denegado. No hay sesión iniciada.\n")
            return False

        if self.usuario_actual.rol not in roles_permitidos:
            print("Acceso denegado. Rol sin permisos suficientes.\n")
            return False

        return True
```
