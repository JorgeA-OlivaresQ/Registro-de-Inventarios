# 2. Módulo de control de stock

## Este módulo permite visualizar el stock de los productos y actualizar existencias.

```python
class ControlStock:
    """
    Clase que permite visualizar y actualizar el stock de productos.
    Recibe como referencia el diccionario de productos.
    """

    def __init__(self, productos):
        self.productos = productos

    def ver_stock_producto(self, codigo):
        try:
            if codigo not in self.productos:
                raise ValueError("No se encontró un producto con ese código.")

            producto = self.productos[codigo]
            print(f"Producto: {producto.nombre} | Stock actual: {producto.cantidad}\n")

        except ValueError as error:
            print(f"Error al consultar stock: {error}\n")

    def actualizar_stock(self, codigo, nueva_cantidad):
        try:
            if codigo not in self.productos:
                raise ValueError("No se encontró un producto con ese código.")

            if nueva_cantidad < 0:
                raise ValueError("La nueva cantidad no puede ser negativa.")

            self.productos[codigo].cantidad = nueva_cantidad
            print("Stock actualizado correctamente.\n")

        except ValueError as error:
            print(f"Error al actualizar stock: {error}\n")

    def mostrar_stock_general(self):
        if not self.productos:
            print("No hay productos registrados en el inventario.\n")
            return

        print("=== STOCK GENERAL DEL INVENTARIO ===")
        for producto in self.productos.values():
            print(f"{producto.codigo} | {producto.nombre} | Stock: {producto.cantidad}")
        print()
```
