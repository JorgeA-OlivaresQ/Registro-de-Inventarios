# 3. Módulo de entradas de inventario

## Este módulo registra el ingreso de nuevos productos o reposición de stock.

```python
class EntradasInventario:
    """
    Clase que registra entradas de inventario.
    """

    def __init__(self, productos):
        self.productos = productos
        self.historial_entradas = []

    def registrar_entrada(self, codigo, cantidad_ingresada, motivo="Reposición"):
        try:
            if codigo not in self.productos:
                raise ValueError("No se encontró un producto con ese código.")

            if cantidad_ingresada <= 0:
                raise ValueError("La cantidad ingresada debe ser mayor que cero.")

            producto = self.productos[codigo]
            producto.cantidad += cantidad_ingresada

            movimiento = {
                "codigo": codigo,
                "nombre": producto.nombre,
                "cantidad_ingresada": cantidad_ingresada,
                "motivo": motivo
            }

            self.historial_entradas.append(movimiento)
            print("Entrada de inventario registrada correctamente.\n")

        except ValueError as error:
            print(f"Error al registrar entrada: {error}\n")

    def mostrar_historial_entradas(self):
        if not self.historial_entradas:
            print("No hay entradas registradas.\n")
            return

        print("=== HISTORIAL DE ENTRADAS ===")
        for entrada in self.historial_entradas:
            print(
                f"Código: {entrada['codigo']} | "
                f"Producto: {entrada['nombre']} | "
                f"Cantidad: {entrada['cantidad_ingresada']} | "
                f"Motivo: {entrada['motivo']}"
            )
        print()
```
