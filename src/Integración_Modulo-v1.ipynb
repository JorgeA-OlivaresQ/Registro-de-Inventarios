# Primera versión del Código completo integrado del sistema de gestión de inventarios

```python
class Usuario:
    """
    Clase que representa a un usuario del sistema.
    """

    def __init__(self, username, password, rol):
        self.username = username
        self.password = password
        self.rol = rol


class AutenticacionUsuarios:
    """
    Clase que administra el registro, inicio de sesión
    y control de acceso según el rol del usuario.
    """

    def __init__(self):
        self.usuarios = {}
        self.usuario_actual = None

    def registrar_usuario(self, username, password, rol):
        try:
            if username in self.usuarios:
                raise ValueError("El nombre de usuario ya existe.")

            if rol not in ["administrador", "almacenero", "consulta"]:
                raise ValueError("Rol no válido.")

            self.usuarios[username] = Usuario(username, password, rol)
            print("Usuario registrado correctamente.\n")

        except ValueError as error:
            print(f"Error al registrar usuario: {error}\n")

    def iniciar_sesion(self, username, password):
        try:
            if username not in self.usuarios:
                raise ValueError("El usuario no existe.")

            usuario = self.usuarios[username]

            if usuario.password != password:
                raise ValueError("Contraseña incorrecta.")

            self.usuario_actual = usuario
            print(f"Bienvenido, {usuario.username}. Rol: {usuario.rol}\n")

        except ValueError as error:
            print(f"Error al iniciar sesión: {error}\n")

    def cerrar_sesion(self):
        if self.usuario_actual:
            print(f"Sesión cerrada para {self.usuario_actual.username}.\n")
            self.usuario_actual = None
        else:
            print("No hay ninguna sesión activa.\n")

    def verificar_acceso(self, roles_permitidos):
        if self.usuario_actual is None:
            print("Acceso denegado. No hay sesión iniciada.\n")
            return False

        if self.usuario_actual.rol not in roles_permitidos:
            print("Acceso denegado. Rol sin permisos suficientes.\n")
            return False

        return True


class Producto:
    """
    Clase que representa un producto dentro del inventario.
    """

    def __init__(self, codigo, nombre, categoria, cantidad, precio):
        self.codigo = codigo
        self.nombre = nombre
        self.categoria = categoria
        self.cantidad = cantidad
        self.precio = precio

    def mostrar_informacion(self):
        return (
            f"Código: {self.codigo} | "
            f"Nombre: {self.nombre} | "
            f"Categoría: {self.categoria} | "
            f"Cantidad: {self.cantidad} | "
            f"Precio: S/ {self.precio:.2f}"
        )


class GestionProductos:
    """
    Clase que administra las operaciones del módulo de gestión de productos.
    """

    def __init__(self):
        self.productos = {}

    def registrar_producto(self, codigo, nombre, categoria, cantidad, precio):
        try:
            if codigo in self.productos:
                raise ValueError("Ya existe un producto con ese código.")
            if cantidad < 0:
                raise ValueError("La cantidad no puede ser negativa.")
            if precio < 0:
                raise ValueError("El precio no puede ser negativo.")

            self.productos[codigo] = Producto(codigo, nombre, categoria, cantidad, precio)
            print("Producto registrado correctamente.\n")

        except ValueError as error:
            print(f"Error al registrar producto: {error}\n")

    def editar_producto(self, codigo, nombre=None, categoria=None, cantidad=None, precio=None):
        try:
            if codigo not in self.productos:
                raise ValueError("No se encontró un producto con ese código.")

            producto = self.productos[codigo]

            if nombre is not None:
                producto.nombre = nombre
            if categoria is not None:
                producto.categoria = categoria
            if cantidad is not None:
                if cantidad < 0:
                    raise ValueError("La cantidad no puede ser negativa.")
                producto.cantidad = cantidad
            if precio is not None:
                if precio < 0:
                    raise ValueError("El precio no puede ser negativo.")
                producto.precio = precio

            print("Producto editado correctamente.\n")

        except ValueError as error:
            print(f"Error al editar producto: {error}\n")

    def eliminar_producto(self, codigo):
        try:
            if codigo not in self.productos:
                raise ValueError("No se encontró un producto con ese código.")

            del self.productos[codigo]
            print("Producto eliminado correctamente.\n")

        except ValueError as error:
            print(f"Error al eliminar producto: {error}\n")

    def consultar_producto(self, codigo):
        try:
            if codigo not in self.productos:
                raise ValueError("No se encontró un producto con ese código.")

            print(self.productos[codigo].mostrar_informacion() + "\n")

        except ValueError as error:
            print(f"Error al consultar producto: {error}\n")

    def listar_productos(self):
        if not self.productos:
            print("No hay productos registrados en el inventario.\n")
            return

        print("\n=== LISTADO DE PRODUCTOS ===")
        for producto in self.productos.values():
            print(producto.mostrar_informacion())
        print()


class ControlStock:
    """
    Clase que permite visualizar y actualizar el stock de productos.
    """

    def __init__(self, productos):
        self.productos = productos

    def ver_stock_producto(self, codigo):
        try:
            if codigo not in self.productos:
                raise ValueError("No se encontró un producto con ese código.")

            producto = self.productos[codigo]
            print(f"Producto: {producto.nombre} | Stock actual: {producto.cantidad}\n")

        except ValueError as error:
            print(f"Error al consultar stock: {error}\n")

    def actualizar_stock(self, codigo, nueva_cantidad):
        try:
            if codigo not in self.productos:
                raise ValueError("No se encontró un producto con ese código.")

            if nueva_cantidad < 0:
                raise ValueError("La nueva cantidad no puede ser negativa.")

            self.productos[codigo].cantidad = nueva_cantidad
            print("Stock actualizado correctamente.\n")

        except ValueError as error:
            print(f"Error al actualizar stock: {error}\n")

    def mostrar_stock_general(self):
        if not self.productos:
            print("No hay productos registrados en el inventario.\n")
            return

        print("\n=== STOCK GENERAL DEL INVENTARIO ===")
        for producto in self.productos.values():
            print(f"{producto.codigo} | {producto.nombre} | Stock: {producto.cantidad}")
        print()


class EntradasInventario:
    """
    Clase que registra entradas de inventario.
    """

    def __init__(self, productos):
        self.productos = productos
        self.historial_entradas = []

    def registrar_entrada(self, codigo, cantidad_ingresada, motivo="Reposición"):
        try:
            if codigo not in self.productos:
                raise ValueError("No se encontró un producto con ese código.")

            if cantidad_ingresada <= 0:
                raise ValueError("La cantidad ingresada debe ser mayor que cero.")

            producto = self.productos[codigo]
            producto.cantidad += cantidad_ingresada

            movimiento = {
                "codigo": codigo,
                "nombre": producto.nombre,
                "cantidad_ingresada": cantidad_ingresada,
                "motivo": motivo
            }

            self.historial_entradas.append(movimiento)
            print("Entrada de inventario registrada correctamente.\n")

        except ValueError as error:
            print(f"Error al registrar entrada: {error}\n")

    def mostrar_historial_entradas(self):
        if not self.historial_entradas:
            print("No hay entradas registradas.\n")
            return

        print("\n=== HISTORIAL DE ENTRADAS ===")
        for entrada in self.historial_entradas:
            print(
                f"Código: {entrada['codigo']} | "
                f"Producto: {entrada['nombre']} | "
                f"Cantidad: {entrada['cantidad_ingresada']} | "
                f"Motivo: {entrada['motivo']}"
            )
        print()


class SalidasInventario:
    """
    Clase que registra salidas de inventario.
    """

    def __init__(self, productos):
        self.productos = productos
        self.historial_salidas = []

    def registrar_salida(self, codigo, cantidad_salida, motivo="Venta"):
        try:
            if codigo not in self.productos:
                raise ValueError("No se encontró un producto con ese código.")

            if cantidad_salida <= 0:
                raise ValueError("La cantidad de salida debe ser mayor que cero.")

            producto = self.productos[codigo]

            if producto.cantidad < cantidad_salida:
                raise ValueError("Stock insuficiente para realizar la salida.")

            producto.cantidad -= cantidad_salida

            movimiento = {
                "codigo": codigo,
                "nombre": producto.nombre,
                "cantidad_salida": cantidad_salida,
                "motivo": motivo
            }

            self.historial_salidas.append(movimiento)
            print("Salida de inventario registrada correctamente.\n")

        except ValueError as error:
            print(f"Error al registrar salida: {error}\n")

    def mostrar_historial_salidas(self):
        if not self.historial_salidas:
            print("No hay salidas registradas.\n")
            return

        print("\n=== HISTORIAL DE SALIDAS ===")
        for salida in self.historial_salidas:
            print(
                f"Código: {salida['codigo']} | "
                f"Producto: {salida['nombre']} | "
                f"Cantidad: {salida['cantidad_salida']} | "
                f"Motivo: {salida['motivo']}"
            )
        print()


def leer_entero(mensaje):
    while True:
        try:
            return int(input(mensaje))
        except ValueError:
            print("Error: debes ingresar un número entero válido.")


def leer_decimal(mensaje):
    while True:
        try:
            return float(input(mensaje))
        except ValueError:
            print("Error: debes ingresar un número válido.")


def mostrar_menu_principal():
    print("=== SISTEMA DE GESTIÓN DE INVENTARIOS ===")
    print("1. Registrar usuario")
    print("2. Iniciar sesión")
    print("3. Cerrar sesión")
    print("4. Registrar producto")
    print("5. Editar producto")
    print("6. Eliminar producto")
    print("7. Consultar producto")
    print("8. Listar productos")
    print("9. Ver stock de un producto")
    print("10. Actualizar stock manualmente")
    print("11. Mostrar stock general")
    print("12. Registrar entrada de inventario")
    print("13. Registrar salida de inventario")
    print("14. Ver historial de entradas")
    print("15. Ver historial de salidas")
    print("16. Salir")


def main():
    auth = AutenticacionUsuarios()
    gestion = GestionProductos()
    stock = ControlStock(gestion.productos)
    entradas = EntradasInventario(gestion.productos)
    salidas = SalidasInventario(gestion.productos)

    while True:
        mostrar_menu_principal()
        opcion = input("Seleccione una opción: ")
        print()

        if opcion == "1":
            username = input("Ingrese nombre de usuario: ")
            password = input("Ingrese contraseña: ")
            rol = input("Ingrese rol (administrador, almacenero, consulta): ")
            auth.registrar_usuario(username, password, rol)

        elif opcion == "2":
            username = input("Ingrese nombre de usuario: ")
            password = input("Ingrese contraseña: ")
            auth.iniciar_sesion(username, password)

        elif opcion == "3":
            auth.cerrar_sesion()

        elif opcion == "4":
            if auth.verificar_acceso(["administrador", "almacenero"]):
                codigo = input("Ingrese el código del producto: ")
                nombre = input("Ingrese el nombre del producto: ")
                categoria = input("Ingrese la categoría del producto: ")
                cantidad = leer_entero("Ingrese la cantidad del producto: ")
                precio = leer_decimal("Ingrese el precio del producto: ")
                gestion.registrar_producto(codigo, nombre, categoria, cantidad, precio)

        elif opcion == "5":
            if auth.verificar_acceso(["administrador", "almacenero"]):
                codigo = input("Ingrese el código del producto a editar: ")
                print("Deje en blanco los campos que no desea modificar.")
                nombre = input("Nuevo nombre: ")
                categoria = input("Nueva categoría: ")
                cantidad_texto = input("Nueva cantidad: ")
                precio_texto = input("Nuevo precio: ")

                cantidad = int(cantidad_texto) if cantidad_texto != "" else None
                precio = float(precio_texto) if precio_texto != "" else None

                gestion.editar_producto(
                    codigo,
                    nombre=nombre if nombre != "" else None,
                    categoria=categoria if categoria != "" else None,
                    cantidad=cantidad,
                    precio=precio
                )

        elif opcion == "6":
            if auth.verificar_acceso(["administrador"]):
                codigo = input("Ingrese el código del producto a eliminar: ")
                gestion.eliminar_producto(codigo)

        elif opcion == "7":
            if auth.verificar_acceso(["administrador", "almacenero", "consulta"]):
                codigo = input("Ingrese el código del producto a consultar: ")
                gestion.consultar_producto(codigo)

        elif opcion == "8":
            if auth.verificar_acceso(["administrador", "almacenero", "consulta"]):
                gestion.listar_productos()

        elif opcion == "9":
            if auth.verificar_acceso(["administrador", "almacenero", "consulta"]):
                codigo = input("Ingrese el código del producto: ")
                stock.ver_stock_producto(codigo)

        elif opcion == "10":
            if auth.verificar_acceso(["administrador", "almacenero"]):
                codigo = input("Ingrese el código del producto: ")
                nueva_cantidad = leer_entero("Ingrese la nueva cantidad: ")
                stock.actualizar_stock(codigo, nueva_cantidad)

        elif opcion == "11":
            if auth.verificar_acceso(["administrador", "almacenero", "consulta"]):
                stock.mostrar_stock_general()

        elif opcion == "12":
            if auth.verificar_acceso(["administrador", "almacenero"]):
                codigo = input("Ingrese el código del producto: ")
                cantidad = leer_entero("Ingrese la cantidad a ingresar: ")
                motivo = input("Ingrese el motivo de la entrada: ")
                entradas.registrar_entrada(codigo, cantidad, motivo)

        elif opcion == "13":
            if auth.verificar_acceso(["administrador", "almacenero"]):
                codigo = input("Ingrese el código del producto: ")
                cantidad = leer_entero("Ingrese la cantidad de salida: ")
                motivo = input("Ingrese el motivo de la salida: ")
                salidas.registrar_salida(codigo, cantidad, motivo)

        elif opcion == "14":
            if auth.verificar_acceso(["administrador", "almacenero", "consulta"]):
                entradas.mostrar_historial_entradas()

        elif opcion == "15":
            if auth.verificar_acceso(["administrador", "almacenero", "consulta"]):
                salidas.mostrar_historial_salidas()

        elif opcion == "16":
            print("Saliendo del sistema...")
            break

        else:
            print("Opción inválida. Intente nuevamente.\n")


if __name__ == "__main__":
    main()
```
