## **A. Código funcional sin automatización y sin buenas prácticas**

```python
class Producto:
    def __init__(self, codigo, nombre, categoria, cantidad, precio):
        self.codigo = codigo
        self.nombre = nombre
        self.categoria = categoria
        self.cantidad = cantidad
        self.precio = precio

    def mostrar_informacion(self):
        return (
            f"Código: {self.codigo} | "
            f"Nombre: {self.nombre} | "
            f"Categoría: {self.categoria} | "
            f"Cantidad: {self.cantidad} | "
            f"Precio: S/ {self.precio:.2f}"
        )


class GestionProductos:
    def __init__(self):
        # Diccionario para almacenar los productos usando el código como clave
        self.productos = {}

    def registrar_producto(self, codigo, nombre, categoria, cantidad, precio):
        try:
            if codigo in self.productos:
                raise ValueError("Ya existe un producto con ese código.")

            if cantidad < 0:
                raise ValueError("La cantidad no puede ser negativa.")

            if precio < 0:
                raise ValueError("El precio no puede ser negativo.")

            nuevo_producto = Producto(codigo, nombre, categoria, cantidad, precio)
            self.productos[codigo] = nuevo_producto
            print("Producto registrado correctamente.")

        except ValueError as error:
            print(f"Error al registrar producto: {error}")

    def editar_producto(self, codigo, nombre=None, categoria=None, cantidad=None, precio=None):
        try:
            if codigo not in self.productos:
                raise ValueError("No se encontró un producto con ese código.")

            producto = self.productos[codigo]

            if nombre is not None:
                producto.nombre = nombre

            if categoria is not None:
                producto.categoria = categoria

            if cantidad is not None:
                if cantidad < 0:
                    raise ValueError("La cantidad no puede ser negativa.")
                producto.cantidad = cantidad

            if precio is not None:
                if precio < 0:
                    raise ValueError("El precio no puede ser negativo.")
                producto.precio = precio

            print("Producto editado correctamente.")

        except ValueError as error:
            print(f"Error al editar producto: {error}")

    def eliminar_producto(self, codigo):
        try:
            if codigo not in self.productos:
                raise ValueError("No se encontró un producto con ese código.")

            del self.productos[codigo]
            print("Producto eliminado correctamente.")

        except ValueError as error:
            print(f"Error al eliminar producto: {error}")

    def consultar_producto(self, codigo):
        try:
            if codigo not in self.productos:
                raise ValueError("No se encontró un producto con ese código.")

            print(self.productos[codigo].mostrar_informacion())

        except ValueError as error:
            print(f"Error al consultar producto: {error}")

    def listar_productos(self):
        if not self.productos:
            print("No hay productos registrados en el inventario.")
            return

        print("\nListado de productos:")
        for producto in self.productos.values():
            print(producto.mostrar_informacion())

# -----------------------------
# Ejemplo de uso del módulo
# -----------------------------
if __name__ == "__main__":
    inventario = GestionProductos()

    # Registrar productos
    inventario.registrar_producto("P001", "Laptop Lenovo", "Tecnología", 10, 2500.00)
    inventario.registrar_producto("P002", "Mouse Logitech", "Accesorios", 25, 80.50)

    # Listar productos
    inventario.listar_productos()

    # Consultar un producto
    inventario.consultar_producto("P001")

    # Editar un producto
    inventario.editar_producto("P002", cantidad=30, precio=85.00)

    # Listar productos actualizados
    inventario.listar_productos()

    # Eliminar un producto
    inventario.eliminar_producto("P001")

    # Listar productos finales
    inventario.listar_productos()
```

## **B. Código funcional y limpio, con buenasprácticas de implementación y manejo de excepciones**

```python
class Producto:
    """
    Clase que representa un producto dentro del inventario.
    """
    def __init__(self, codigo, nombre, categoria, cantidad, precio):
        self.codigo = codigo
        self.nombre = nombre
        self.categoria = categoria
        self.cantidad = cantidad
        self.precio = precio

    def mostrar_informacion(self):
        """
        Devuelve la información del producto en formato de texto.
        """
        return (
            f"Código: {self.codigo} | "
            f"Nombre: {self.nombre} | "
            f"Categoría: {self.categoria} | "
            f"Cantidad: {self.cantidad} | "
            f"Precio: S/ {self.precio:.2f}"
        )


class GestionProductos:
    """
    Clase que administra las operaciones del módulo de gestión de productos.
    """
    def __init__(self):
        self.productos = {}

    def registrar_producto(self, codigo, nombre, categoria, cantidad, precio):
        """
        Registra un nuevo producto en el inventario.
        """
        try:
            if codigo in self.productos:
                raise ValueError("Ya existe un producto con ese código.")
            if cantidad < 0:
                raise ValueError("La cantidad no puede ser negativa.")
            if precio < 0:
                raise ValueError("El precio no puede ser negativo.")

            self.productos[codigo] = Producto(codigo, nombre, categoria, cantidad, precio)
            print("Producto registrado correctamente.\n")

        except ValueError as error:
            print(f"Error al registrar producto: {error}\n")

    def editar_producto(self, codigo, nombre=None, categoria=None, cantidad=None, precio=None):
        """
        Edita los datos de un producto existente.
        Solo actualiza los campos enviados.
        """
        try:
            if codigo not in self.productos:
                raise ValueError("No se encontró un producto con ese código.")

            producto = self.productos[codigo]

            if nombre:
                producto.nombre = nombre
            if categoria:
                producto.categoria = categoria
            if cantidad is not None:
                if cantidad < 0:
                    raise ValueError("La cantidad no puede ser negativa.")
                producto.cantidad = cantidad
            if precio is not None:
                if precio < 0:
                    raise ValueError("El precio no puede ser negativo.")
                producto.precio = precio

            print("Producto editado correctamente.\n")

        except ValueError as error:
            print(f"Error al editar producto: {error}\n")

    def eliminar_producto(self, codigo):
        """
        Elimina un producto del inventario.
        """
        try:
            if codigo not in self.productos:
                raise ValueError("No se encontró un producto con ese código.")

            del self.productos[codigo]
            print("Producto eliminado correctamente.\n")

        except ValueError as error:
            print(f"Error al eliminar producto: {error}\n")

    def consultar_producto(self, codigo):
        """
        Consulta un producto específico por su código.
        """
        try:
            if codigo not in self.productos:
                raise ValueError("No se encontró un producto con ese código.")

            print(self.productos[codigo].mostrar_informacion() + "\n")

        except ValueError as error:
            print(f"Error al consultar producto: {error}\n")

    def listar_productos(self):
        """
        Muestra todos los productos registrados en el inventario.
        """
        if not self.productos:
            print("No hay productos registrados en el inventario.\n")
            return

        print("\nListado de productos:")
        for producto in self.productos.values():
            print(producto.mostrar_informacion())
        print()


def mostrar_menu():
    print("=== SISTEMA DE GESTIÓN DE PRODUCTOS ===")
    print("1. Registrar producto")
    print("2. Editar producto")
    print("3. Eliminar producto")
    print("4. Consultar producto")
    print("5. Listar productos")
    print("6. Salir")


def leer_entero(mensaje):
    while True:
        try:
            return int(input(mensaje))
        except ValueError:
            print("Error: debes ingresar un número entero válido.")


def leer_decimal(mensaje):
    while True:
        try:
            return float(input(mensaje))
        except ValueError:
            print("Error: debes ingresar un número válido.")


def main():
    inventario = GestionProductos()

    while True:
        mostrar_menu()
        opcion = input("Seleccione una opción: ")
        print()

        if opcion == "1":
            codigo = input("Ingrese el código del producto: ")
            nombre = input("Ingrese el nombre del producto: ")
            categoria = input("Ingrese la categoría del producto: ")
            cantidad = leer_entero("Ingrese la cantidad del producto: ")
            precio = leer_decimal("Ingrese el precio del producto: ")
            inventario.registrar_producto(codigo, nombre, categoria, cantidad, precio)

        elif opcion == "2":
            codigo = input("Ingrese el código del producto a editar: ")

            print("Deje en blanco los campos que no desea modificar.")
            nombre = input("Nuevo nombre: ")
            categoria = input("Nueva categoría: ")
            cantidad_texto = input("Nueva cantidad: ")
            precio_texto = input("Nuevo precio: ")

            cantidad = int(cantidad_texto) if cantidad_texto != "" else None
            precio = float(precio_texto) if precio_texto != "" else None

            inventario.editar_producto(
                codigo,
                nombre=nombre if nombre != "" else None,
                categoria=categoria if categoria != "" else None,
                cantidad=cantidad,
                precio=precio
            )

        elif opcion == "3":
            codigo = input("Ingrese el código del producto a eliminar: ")
            inventario.eliminar_producto(codigo)

        elif opcion == "4":
            codigo = input("Ingrese el código del producto a consultar: ")
            inventario.consultar_producto(codigo)

        elif opcion == "5":
            inventario.listar_productos()

        elif opcion == "6":
            print("Saliendo del sistema...")
            break

        else:
            print("Opción inválida. Intente nuevamente.\n")


if __name__ == "__main__":
    main()
```
